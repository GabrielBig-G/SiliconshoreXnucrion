The missing values should be handled by filling them... 

Now I determined that I can fill Monthly_Income and Income_Copy with median because income is usually highly skewed and I don't want outliers affecting the column by using mean and mode, so it's best I use median that just picks inbetween... 

I also decide to fill Credit_Score with median because I'm trying to prevent outliers... 

I decide to fill Owns_House with mode because it's a Boolean column and the best approach to Boolean columns is mode...

```python
# Filling Monthly_Income with median
df['Monthly_Income'] = df['Monthly_Income'].fillna(df['Monthly_Income'].median())

# Filling Income_Copy with median
df['Income_Copy'] = df['Income_Copy'].fillna(df['Income_Copy'].median())

# Filling Credit_Score with median
df['Credit_Score'] = df['Credit_Score'].fillna(df['Credit_Score'].median())

# Filling Own_House with mode
df['Owns_House'] = df['Owns_House'].fillna(df['Owns_House'].mode()[0])
```

# Part 5: Feature Scaling

# 5.1 

Scaler Necessity: Features like "Monthly_Income" reach into millions, while "Age" values hover under 60. Left unscaled, distance-based models assume income is vastly more important than age simply due to its unit magnitude.

What StandardScaler does: It centers the column distribution around \(0\) with unit variance.

Mean destination: It shifts exactly to 0.

Standard Deviation destination: It scales exactly to 1.

Features to standardize: Age, Credit_Score, Years_Employed, Num_Children, Monthly_Transport, Satisfaction_Score. These are the features without extreme skew or massive outliers.

```python
Scaler = StandardScaler()

#Age StandardScaler
Scaler.set_output(transform="pandas")
StandardAge = Scaler.fit_transform(df[['Age']])
df['Age'] = StandardAge

#Credit_Score StandardScaler
Scaler.set_output(transform="pandas")
StandardCredit_Score = Scaler.fit_transform(df[['Credit_Score']])
df['Credit_Score'] = StandardCredit_Score

#Years_Employed StandardScaler
Scaler.set_output(transform="pandas")
StandardYears_Employed = Scaler.fit_transform(df[['Years_Employed']])
df['Years_Employed'] = StandardYears_Employed

#Num_Children StandardScaler
Scaler.set_output(transform="pandas")
StandardNum_Children = Scaler.fit_transform(df[['Num_Children']])
df['Num_Children'] = StandardNum_Children

#Monthly_Transport StandardScaler
Scaler.set_output(transform="pandas")
StandardMonthly_Transport = Scaler.fit_transform(df[['Monthly_Transport']])
df['Monthly_Transport'] = StandardMonthly_Transport

#Satisfaction_Score StandardScaler
Scaler.set_output(transform="pandas")
StandardSatisfaction_Score = Scaler.fit_transform(df[['Satisfaction_Score']])
df['Satisfaction_Score'] = StandardSatisfaction_Score

df
```

# 5.2

MinMaxScalerRange: Transforms data precisely into a [0, 1] bound.

Usefulness: Excellent for algorithms that require defined lower and upper bounds, such as neural networks or image processing setups.

Extreme value impact: Outliers compress all normal records into a tiny, indistinguishable band close to zero.

Affected Features: Monthly_Income, Loan_Amount, and Savings_Balance are heavily compressed due to the massive outlier present in customer record C018

```python
Scaler = MinMaxScaler()

# Monthly_Income MinMaxScaler
Scaler.set_output(transform="pandas")
MinMaxMonthly_Income = Scaler.fit_transform(df[['Monthly_Income']])
df['Monthly_Income'] = MinMaxMonthly_Income

# Loan_Amount MinMaxScaler
Scaler.set_output(transform="pandas")
MinMaxLoan_Amount = Scaler.fit_transform(df[['Loan_Amount']])
df['Loan_Amount'] = MinMaxLoan_Amount

# Savings_Balance MinMaxScaler
Scaler.set_output(transform="pandas")
MinMaxSavings_Balance = Scaler.fit_transform(df[['Savings_Balance']])
df['Savings_Balance'] = MinMaxSavings_Balance

df
```

# 5.3

Difference: It uses the Median and Interquartile Range (IQR) for scaling instead of the Mean and Standard Deviation as StandardScaler does.

Outlier Utility: Because the median and IQR ignore extreme trailing tails, outliers do not warp the scaling of regular customer records.

Extreme Value Columns: Monthly_Income, Loan_Amount, and Savings_Balance.

Most Appropriate Scaler: RobustScaler is the most optimal choice for these specific heavily skewed features.

Universal application rule: No, RobustScaler should not be used universally. For features with standard normal distributions, StandardScaler provides clearer signals for variance preservation.

# Part 6: Choose the appropriate Scaler

Situation 1 (Normal distribution, no outliers): StandardScaler. Preserves the shape and variance profile cleanly.

Situation 2 (Extreme observations present): RobustScaler. Neutralizes outlier distortion using medians and IQRs.

Situation 3 (Constrained between 0 and 1): MinMaxScaler. Algebraically enforces hard minimum/maximum boundaries.

Situation 4 (Decision Tree): No Scaler Required. Decision trees branch by splitting features individually; monotonic transformations do not change the split points.

Situation 5 (K-Nearest Neighbours): StandardScaler / RobustScaler. Crucial because KNN uses Euclidean geometric distance metrics.

Situation 6 (K-Means clustering): StandardScaler. Equalizes scale parameters so all dimensions contribute equally to cluster centroids.

Situation 7 (Neural network): MinMaxScaler / StandardScaler. Stabilizes gradient descent optimization and prevents exploding weight states.
